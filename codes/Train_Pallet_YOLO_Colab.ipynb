### 3. Extract Dataset (`dataset_for_colab.zip`)
Upload file dataset_for_colab.zip ke panel file Colab di sebelah kiri.

```python
import os
import zipfile

zip_file_path = '/content/dataset_for_colab.zip'
if not os.path.exists(zip_file_path):
    zip_file_path = '/content/drive/MyDrive/dataset_for_colab.zip'

if os.path.exists(zip_file_path):
    with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
        zip_ref.extractall('/content/dataset')
    print('Dataset berhasil di-extract ke /content/dataset!')
else:
    print('File dataset_for_colab.zip tidak ditemukan! Tolong upload file tersebut terlebih dahulu.')
```

### 4. Mulai Training AI (YOLOv8)

```python
from ultralytics import YOLO

# Load pretrained YOLOv8 nano model
model = YOLO('yolov8n.pt')

# Mulai training dengan GPU Colab
results = model.train(
    data='/content/dataset/data.yaml',
    epochs=100,
    imgsz=640,
    batch=16,
    device=0,
    project='pallet_colab_run',
    name='variant_classifier_v3',
    exist_ok=True
)
print('Training Selesai!')
```

### 5. Download Model Hasil Training (`best.pt`)

```python
from google.colab import files
import os

best_model_path = '/content/pallet_colab_run/variant_classifier_v3/weights/best.pt'
if os.path.exists(best_model_path):
    files.download(best_model_path)
    print('Menyiapkan download best.pt...')
else:
    print('File best.pt belum ditemukan!')
```
